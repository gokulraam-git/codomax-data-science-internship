```python
players = [
    {
        "name": "Virat Kohli",
        "matches": 10,
        "runs": 520,
        "wickets": 2
    },
    {
        "name": "Rohit Sharma",
        "matches": 10,
        "runs": 450,
        "wickets": 1
    },
    {
        "name": "Rishabh Pant",
        "matches": 10,
        "runs": 390,
        "wickets": 0
    },
    {
        "name": "Hardik Pandya",
        "matches": 10,
        "runs": 280,
        "wickets": 12
    },
    {
        "name": "Jasprit Bumrah",
        "matches": 10,
        "runs": 80,
        "wickets": 18
    }
]

print("Number of Players:", len(players))
```

```python
for player in players:
    print(player["name"], "->", player["runs"], "runs")
```

```python
for player in players:
    runs_per_match = player["runs"] / player["matches"]

    print(
        player["name"],
        "->",
        round(runs_per_match, 2),
        "runs/match"
    )
```

```python
def calculate_runs_per_match(runs, matches):
    return runs / matches

runs = 520
matches = 10

result = calculate_runs_per_match(runs, matches)

print("Runs per Match:", result)
```

```python
def performance_level(runs):
    if runs >= 500:
        return "Excellent"
    elif runs >= 300:
        return "Good"
    elif runs >= 200:
        return "Average"
    else:
        return "Needs Improvement"

for player in players:
    level = performance_level(player["runs"])
    print(player["name"], "->", level)
```

```python
top_batsman = max(players, key=lambda player: player["runs"])

print(" Top Batsman:", top_batsman["name"])
print("Runs:", top_batsman["runs"])
```

```python
best_bowler = max(players, key=lambda player: player["wickets"])

print(" Best Bowler:", best_bowler["name"])
print("Wickets:", best_bowler["wickets"])
```

```python
with open("cricket_players.txt", "w") as file:
    for player in players:
        file.write(
            f"{player['name']}, "
            f"Matches: {player['matches']}, "
            f"Runs: {player['runs']}, "
            f"Wickets: {player['wickets']}\n"
        )

print(" Cricket player data saved successfully!")
```

#  What is Data Science?

Data Science is the process of collecting, cleaning, analyzing
and interpreting data to extract useful information and insights.

### Data Science Process:

1. Data Collection
2. Data Cleaning
3. Data Analysis
4. Data Visualization
5. Machine Learning
6. Decision Making

### Cricket Applications:

- Player performance analysis
- Match prediction
- Team selection
- Player ranking
- Injury analysis
- Win probability prediction
- Bowling and batting analysis

```python
print("=" * 50)
print(" CRICKET PLAYER PERFORMANCE ANALYSIS")
print("=" * 50)

for player in players:
    runs_per_match = player["runs"] / player["matches"]

    print("\nPlayer:", player["name"])
    print("Matches:", player["matches"])
    print("Runs:", player["runs"])
    print("Wickets:", player["wickets"])
    print("Runs per Match:", round(runs_per_match, 2))
```

```python
total_runs = sum(player["runs"] for player in players)
total_wickets = sum(player["wickets"] for player in players)

average_runs = total_runs / len(players)

print(" OVERALL STATISTICS")
print("-" * 30)

print("Total Runs:", total_runs)
print("Total Wickets:", total_wickets)
print("Average Runs per Player:", round(average_runs, 2))
```

```python
top_batsman = max(players, key=lambda player: player["runs"])
best_bowler = max(players, key=lambda player: player["wickets"])

print(" TOP BATSMAN")
print("Name:", top_batsman["name"])
print("Runs:", top_batsman["runs"])

print("\n BEST BOWLER")
print("Name:", best_bowler["name"])
print("Wickets:", best_bowler["wickets"])
```

```python
print(" PERFORMANCE SUMMARY")
print("=" * 50)

for player in players:
    performance = performance_level(player["runs"])

    print(
        f"{player['name']} | "
        f"Runs: {player['runs']} | "
        f"Wickets: {player['wickets']} | "
        f"Performance: {performance}"
    )
```
