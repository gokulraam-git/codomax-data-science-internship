```python
top_run_scorers = df.sort_values(by="Runs", ascending=False).head(5)

top_run_scorers[["Player_Name", "Team", "Runs"]]
```

```python
high_average_players = df[df["Batting_Average"] > 40]

high_average_players[
    ["Player_Name", "Team", "Batting_Average"]
]
```

```python
team_average_runs = df.groupby("Team")["Runs"].mean()

team_average_runs.sort_values(ascending=False)
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
df[df.isnull().any(axis=1)]
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("Missing numerical values have been handled.")
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

df = df.reset_index(drop=True)

print("Duplicate rows after removal:", df.duplicated().sum())
```

```python
print("Rows after cleaning:", df.shape[0])
print("Columns after cleaning:", df.shape[1])
```

```python
print(df.dtypes)
```

```python
runs = np.array(df["Runs"])
wickets = np.array(df["Wickets"])

print("Runs Array:")
print(runs)

print("\nWickets Array:")
print(wickets)
```

```python
print("Mean Runs:", np.mean(runs))
print("Median Runs:", np.median(runs))
print("Maximum Runs:", np.max(runs))
print("Minimum Runs:", np.min(runs))
print("Standard Deviation:", np.std(runs))
```

```python
print("Mean Wickets:", np.mean(wickets))
print("Maximum Wickets:", np.max(wickets))
print("Minimum Wickets:", np.min(wickets))
```

```python
top_batsman = df.loc[df["Runs"].idxmax()]

print("Top Run Scorer")
print("----------------")
print("Player:", top_batsman["Player_Name"])
print("Team:", top_batsman["Team"])
print("Runs:", top_batsman["Runs"])
```

```python
top_bowler = df.loc[df["Wickets"].idxmax()]

print("Top Wicket Taker")
print("----------------")
print("Player:", top_bowler["Player_Name"])
print("Team:", top_bowler["Team"])
print("Wickets:", top_bowler["Wickets"])
```

```python
best_average = df.loc[df["Batting_Average"].idxmax()]

print("Highest Batting Average")
print("-----------------------")
print("Player:", best_average["Player_Name"])
print("Team:", best_average["Team"])
print("Batting Average:", best_average["Batting_Average"])
```
