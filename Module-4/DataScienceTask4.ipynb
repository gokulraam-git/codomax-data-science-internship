## 🤖 Machine Learning Approach

This project uses **Supervised Learning** because the model learns from
existing cricket player data where the target value (Runs) is already known.

### Algorithm Used

**Linear Regression**

Linear Regression is a simple Machine Learning algorithm used to predict
a numerical value based on one or more input features.

### Prediction Task

**Input Features → Player Performance Statistics**

**Target → Runs**

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("Libraries imported successfully!")
```

## 📂 Load Dataset

The cricket player performance dataset is loaded using Pandas.

```python
df = pd.read_csv("cricket_player_performance.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
df.head()
```

## 🔍 Dataset Exploration

Before building the Machine Learning model, the dataset is examined
to understand its structure and identify missing values and duplicate records.

```python
print("Column Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

## 🧹 Data Cleaning

Machine Learning models require clean numerical data.

The following cleaning steps are performed:

- Missing numerical values are replaced using the median.
- Duplicate records are removed.
- The dataset is reset with a clean index.

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

df = df.drop_duplicates()
df = df.reset_index(drop=True)

print("Data cleaning completed!")
print("Dataset shape after cleaning:", df.shape)
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
```

## 🎯 Feature Selection

The following player performance statistics are selected as input features:

- Matches
- Batting Average
- Strike Rate
- Fours
- Sixes

### Target Variable

**Runs**

The model will learn the relationship between these performance
statistics and total runs.

```python
features = [
    "Matches",
    "Batting_Average",
    "Strike_Rate",
    "Fours",
    "Sixes"
]

X = df[features]
y = df["Runs"]

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

## ✂️ Train-Test Split

The dataset is divided into two parts:

- **80% Training Data** – Used to train the Machine Learning model
- **20% Testing Data** – Used to evaluate the model on unseen data

A fixed random state is used so that the same split can be reproduced.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 🧠 Model Training

A **Linear Regression** model from Scikit-learn is used.

The model learns patterns from the training data and creates a relationship
between the selected player statistics and runs.

```python
model = LinearRegression()

model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
```

## 🔮 Making Predictions

The trained model is now used to predict runs for the testing dataset.

```python
y_pred = model.predict(X_test)

print("Predicted Runs:")
print(y_pred[:10])
```

## 📊 Model Evaluation

The model is evaluated using three common regression metrics:

### Mean Absolute Error (MAE)
Measures the average absolute difference between actual and predicted values.

### Mean Squared Error (MSE)
Measures the average squared difference between actual and predicted values.

### R² Score
Shows how well the model explains the variation in the target variable.

For MAE and MSE, lower values generally indicate smaller prediction errors.
For R², a value closer to 1 indicates that the model explains more of the
variation in the target data.

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Model Evaluation")
print("----------------")
print("Mean Absolute Error (MAE):", round(mae, 2))
print("Mean Squared Error (MSE):", round(mse, 2))
print("R² Score:", round(r2, 2))
```

## 📋 Actual vs Predicted Runs

The actual and predicted values are compared to understand how the model
performs on individual test samples.

```python
comparison = pd.DataFrame({
    "Actual Runs": y_test.values,
    "Predicted Runs": np.round(y_pred, 2)
})

comparison.head(10)
```

## 🧪 Sample Player Prediction

The trained model can also be used to predict the runs of a new player
based on their performance statistics.

```python
sample_player = pd.DataFrame({
    "Matches": [100],
    "Batting_Average": [40],
    "Strike_Rate": [100],
    "Fours": [200],
    "Sixes": [50]
})

predicted_runs = model.predict(sample_player)

print("Sample Player Statistics:")
print(sample_player)

print("\nPredicted Runs:", round(predicted_runs[0], 2))
```
