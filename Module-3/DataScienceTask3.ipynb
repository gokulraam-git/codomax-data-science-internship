## Objectives

- Load and explore the cricket dataset
- Create different visualizations using Matplotlib and Seaborn
- Create Bar, Line, Pie, Histogram, and Scatter plots
- Understand patterns and trends in the data
- Identify useful insights from cricket player statistics

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

## Load the Dataset

The cricket player performance dataset is loaded using Pandas.

```python
df = pd.read_csv("cricket_player_performance.csv")

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
```

## 1. Bar Chart – Top 10 Run Scorers

A bar chart is used to compare the runs scored by the top 10 players.

```python
top_10_runs = df.sort_values(by="Runs", ascending=False).head(10)

plt.figure(figsize=(10, 5))
plt.bar(top_10_runs["Player_Name"], top_10_runs["Runs"])

plt.title("Top 10 Run Scorers")
plt.xlabel("Player")
plt.ylabel("Runs")
plt.xticks(rotation=45)

plt.show()
```

### Observation

The bar chart makes it easy to compare the run-scoring performance of the top 10 players.
Players with taller bars have scored more runs.

## 2. Line Chart – Matches and Runs

A line chart is used to observe the relationship between matches played and runs scored.

```python
line_data = df.sort_values(by="Matches").head(15)

plt.figure(figsize=(10, 5))
plt.plot(line_data["Matches"], line_data["Runs"], marker="o")

plt.title("Matches Played vs Runs Scored")
plt.xlabel("Matches")
plt.ylabel("Runs")
plt.grid(True)

plt.show()
```

### Observation

The line chart shows how runs vary among players with different numbers of matches.
It helps identify general changes in runs as matches played increase.

## 3. Pie Chart – Players by Team

A pie chart is used to show how the players are distributed across different teams.

```python
team_counts = df["Team"].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(
    team_counts.values,
    labels=team_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Player Distribution by Team")

plt.show()
```

### Observation

The pie chart shows the proportion of players belonging to each team.
It provides a quick view of the team-wise distribution in the dataset.

## 4. Histogram – Distribution of Runs

A histogram is used to understand how the runs are distributed among the players.

```python
plt.figure(figsize=(10, 5))

plt.hist(df["Runs"], bins=10, edgecolor="black")

plt.title("Distribution of Runs")
plt.xlabel("Runs")
plt.ylabel("Number of Players")

plt.show()
```

### Observation

The histogram shows how frequently different ranges of runs occur.
It helps understand the distribution of player run-scoring performance.

## 5. Scatter Plot – Runs vs Batting Average

A scatter plot is used to observe the relationship between runs scored and batting average.

```python
plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=df,
    x="Batting_Average",
    y="Runs"
)

plt.title("Runs vs Batting Average")
plt.xlabel("Batting Average")
plt.ylabel("Runs")

plt.show()
```

### Observation

The scatter plot shows the relationship between batting average and total runs.
It helps identify players with high runs and high batting averages.

## 🔍 Key Insights
